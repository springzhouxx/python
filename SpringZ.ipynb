```python
def calc_total_bill (price, tax_rate=0.13, tip_rate=0.15):
    tax_amt = price * tax_rate
    tip_amt = price * tip_rate
    return price + tax_amt + tip_amt
```

```python
def eye_exam_covered(age, qualifying_condition, time_since_last_exam):
    '''assume (1) age is paased as an int is for the age of a person 
    (2) qualifying_condition is passed as a bool (True/False) 
    indiciating whether a eligible medical condition is met, 
    and (3) tune_since_last_exam is passed as int for 
    the number of months since the last eye exam'''
    if 0 <= age <= 19:
        if time_since_last_exam >= 12:
            return "You are eligible for one major eye exam and any minor assessment."
        else:
            return "It hasn't been 12 months since your last major eye exam."
    elif 20<= age <=64:
        if qualifying_condition: #if qualifying_condition == True is also fine but just better to leave it blank
            if time_since_last_exam >=12:
                return "You are eligibale for 1 major eye exam and and two additional follow-up minor assessment."
            else:
                return "It hasn't been 12 months since your last major eye exam."
        else:
            return "You do not have a qualifying eligible medical condition."
    elif age >= 65:
        if time_since_last_exam >= 18:
            return "You are eligibale for 1 major eye exam and and two additional follow-up minor assessment."
        else:
            return "It hasn't been 12 months since your last major eye exam."
    else:
        return "Invalid age input"
```

```python
rainbow = {"red", "orange", "yellow", "green", "blue", "indigo", "violet"}
olympic_flag = {"red", "green", "yellow", "blue", "black"}
```

```python
# doff: find the value in the first set that are not in the second set
rainbow.difference(olympic_flag)
```

```python
# symetric diff: find the value that are in only one of the two sets
rainbow.symmetric_difference(olympic_flag)
```

```python
# intersection: find the values the two sets are in common
rainbow.intersection(olympic_flag)
```

```python
# union: combine the two sets to get the unique values in both
rainbow.union(olympic_flag)
```
